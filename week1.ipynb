```python
# --- Task 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# 1.1 Generate a list of 100 simulated sensor readings.
#    Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value).
#    - timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
#    - device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
#    - device_type: Randomly chosen from DEVICE_TYPES.
#    - location: Randomly chosen from LOCATIONS.
#    - value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
# Store these in a list called 'raw_sensor_data'.
import random
from datetime import datetime, timedelta

DEVICE_TYPES = ["Temperature", "Humidity", "Pressure", "Light", "CO2"]

LOCATIONS = [
    "Lab A",
    "Lab B",
    "Warehouse",
    "Office",
    "Factory Floor"
]

READING_RANGES = {
    "Temperature": (15.0, 40.0),   # °C
    "Humidity": (20.0, 90.0),      # %
    "Pressure": (950.0, 1050.0),   # hPa
    "Light": (100.0, 1000.0),      # lux
    "CO2": (300.0, 2000.0)         # ppm
}

# Base timestamp
start_time = datetime(2026, 1, 1, 0, 0, 0)

raw_sensor_data = []

for i in range(100):
    timestamp = (start_time + timedelta(minutes=i)).strftime("%Y-%m-%d %H:%M:%S")
    device_id = f"DEV{i+1:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)
    value = round(random.uniform(*READING_RANGES[device_type]), 2)

    raw_sensor_data.append(
        (timestamp, device_id, device_type, location, value)
    )

# Example: print the generated readings
for reading in raw_sensor_data:
    print(reading)
```

```python
# 1.2 From 'raw_sensor_data', create a list named 'temperature_readings' containing only the readings
#     where 'device_type' is 'Temperature_Sensor'.
#     Each item in 'temperature_readings' should be a dictionary with keys:
#     'timestamp', 'device_id', 'location', 'temperature_celsius'.
# Step 1: Create the raw sensor data
raw_sensor_data = [
    ("2026-07-27 10:00:00", "DEV001", "Temperature_Sensor", "Room_A", 25.6),
    ("2026-07-27 10:01:00", "DEV002", "Humidity_Sensor", "Room_B", 60.2),
    ("2026-07-27 10:02:00", "DEV003", "Temperature_Sensor", "Room_C", 24.8)
]

# Step 2: Create an empty list to store temperature readings
temperature_readings = []

# Step 3: Go through each sensor reading
for timestamp, device_id, device_type, location, value in raw_sensor_data:

    # Check if the device is a Temperature Sensor
    if device_type == "Temperature_Sensor":

        # Create a dictionary and add it to the list
        reading = {
            "timestamp": timestamp,
            "device_id": device_id,
            "location": location,
            "temperature_celsius": value
        }

        temperature_readings.append(reading)

# Step 4: Display the temperature readings
print("Temperature Readings:")

for reading in temperature_readings:
    print(reading)
```

```python
# 1.4 Create a dictionary named 'device_stats' where keys are 'device_type' and values are
#     a tuple: (min_value, max_value) observed for that device type across all readings.
#     Initialize with sensible large/small values.
# Step 1: Create the raw sensor data
raw_sensor_data = [
    ("2026-07-27 10:00:00", "DEV001", "Temperature_Sensor", "Room_A", 25.6),
    ("2026-07-27 10:01:00", "DEV002", "Humidity_Sensor", "Room_B", 60.2),
    ("2026-07-27 10:02:00", "DEV003", "Temperature_Sensor", "Room_C", 24.8),
    ("2026-07-27 10:03:00", "DEV004", "Humidity_Sensor", "Room_A", 58.4)
]

# Step 2: Create an empty dictionary
device_stats = {}

# Step 3: Go through each reading
for timestamp, device_id, device_type, location, value in raw_sensor_data:

    # If the device type is seen for the first time
    if device_type not in device_stats:
        device_stats[device_type] = (value, value)

    # Otherwise, update the minimum and maximum values
    else:
        min_value, max_value = device_stats[device_type]

        if value < min_value:
            min_value = value

        if value > max_value:
            max_value = value

        device_stats[device_type] = (min_value, max_value)

# Step 4: Display the dictionary
print(device_stats)
```

```python
# --- Task 2: Writing and Using Functions in Python ---
print("\n--- Task 2: Function Definitions and Usage ---")

# 2.1 Write a function `get_readings_by_location(readings_list, target_location)`
#     that takes the 'raw_sensor_data' list and a 'target_location' string.
#     It should return a list of all readings (as tuples) for that specific location.
# Function to get all readings for a specific location
def get_readings_by_location(readings_list, target_location):

    # Create an empty list to store matching readings
    location_readings = []

    # Go through each reading
    for reading in readings_list:

        # Check if the location matches
        if reading[3] == target_location:
            location_readings.append(reading)

    # Return the list of matching readings
    return location_readings


# Example raw sensor data
raw_sensor_data = [
    ("2026-07-27 10:00:00", "DEV001", "Temperature_Sensor", "Room_A", 25.6),
    ("2026-07-27 10:01:00", "DEV002", "Humidity_Sensor", "Room_B", 60.2),
    ("2026-07-27 10:02:00", "DEV003", "Temperature_Sensor", "Room_A", 24.8),
    ("2026-07-27 10:03:00", "DEV004", "Pressure_Sensor", "Room_C", 101.3)
]

# Call the function
result = get_readings_by_location(raw_sensor_data, "Room_A")

# Print the result
print(result)
```

```python
# 2.2 Write a function `calculate_average_temperature(temp_data_list)`
#     that takes the 'temperature_readings' list (list of dictionaries).
#     It should calculate and return the average temperature. Return 0.0 if the list is empty.

temperature_readings = [
    {
        "timestamp": "2026-07-27 10:00:00",
        "device_id": "DEV001",
        "location": "Room_A",
        "temperature_celsius": 25.6
    },
    {
        "timestamp": "2026-07-27 10:02:00",
        "device_id": "DEV003",
        "location": "Room_C",
        "temperature_celsius": 24.8
    }
]
# Function to calculate average temperature
def calculate_average_temperature(temp_data_list):

    # If the list is empty, return 0.0
    if len(temp_data_list) == 0:
        return 0.0

    # Variable to store the total temperature
    total_temperature = 0

    # Loop through each dictionary
    for reading in temp_data_list:
        total_temperature += reading["temperature_celsius"]

    # Calculate the average
    average = total_temperature / len(temp_data_list)

    return average


# Call the function
average_temp = calculate_average_temperature(temperature_readings)

print("Average Temperature:", average_temp)
```

```python
# 2.3 Write a function `generate_report_line(reading_dict)`
#     that takes a single temperature reading dictionary.
#     It should return a formatted string like:
#     "[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C"

# Function to generate a report line
def generate_report_line(reading_dict):

    report = f"[{reading_dict['timestamp']}] Device {reading_dict['device_id']} at {reading_dict['location']}: {reading_dict['temperature_celsius']}°C"

    return report


# Example dictionary
reading = {
    "timestamp": "2026-07-27 10:00:00",
    "device_id": "DEV001",
    "location": "Room_A",
    "temperature_celsius": 25.6
}

# Call the function
result = generate_report_line(reading)

print(result)
```

```python
# 3.2 Calculate the duration between the first and last timestamp in 'raw_sensor_data'.
#     Convert string timestamps to datetime objects first.
#     Print the duration in days, hours, and minutes.

# Your code for Task 3.2 here:
from datetime import datetime

# Convert string timestamps to datetime objects
timestamps = []
for reading in raw_sensor_data:
    timestamp = datetime.strptime(reading[0], "%Y-%m-%d %H:%M:%S")
    timestamps.append(timestamp)

# Calculate the duration between the first and last timestamp
duration = timestamps[-1] - timestamps[0]

# Extract days, hours, and minutes
days = duration.days
hours = duration.seconds // 3600
minutes = (duration.seconds % 3600) // 60

# Print the result
print("Duration:")
print("Days:", days)
print("Hours:", hours)
print("Minutes:", minutes)
```
